```python
import numpy as np
import astropy as c
```

```python
#Initial data
diameter= 0.5 #[m] diameter of telescopes aperture
q= 7.2312 *10**-7 #[J] in 2 min
d_q =0.0064 *10**-7 #Uncertainty of q
prlx=90.91 #[m.a.s]
d_prlx=0.52 #Uncertainty of parallax
t_eff=8262 #[K]
d_t= 36 #uncertainty of temperature
```

```python
#1: Distance to the Star
print("=======================================")
print("1: Distance to the Star:")
print("---------------------------------------")
print("Analitical Propagation:")
dist= 1/(prlx*1e-3) #[pc]
d_dist = 1/(prlx*1e-3)**2 * (d_prlx*1e-3)
print("Distance = {:2.4e} +/- {:2.4e} pc".format(dist, d_dist))
print("Relative Error = {:2.4e} %".format(d_dist/dist*100))
print("--------------------------------------")
print("Numerical Propagation:")
num_samples=1000
parallax_simul= prlx + d_prlx * np.random.randn(num_samples)
print("Parallax = {:2.4e} +/- {:2.4e} pc".format(np.mean(parallax_simul), np.std(parallax_simul)))
dist_simul= 1/(parallax_simul*1e-3)
mean_dist=np.mean(dist_simul) 
std_dist = np.std(dist_simul-mean_dist)
print("Distance = {:2.4e} +/- {:2.4e} pc".format(mean_dist, std_dist))
print("Relative Error = {:2.4e} %".format(std_dist/mean_dist*100))
```

```python
print("=======================================")
print("2: Irradiance at the telescope's aperture:")
print("---------------------------------------")
print("Analitical Propagation:")
print("Radiant energy = {:2.4e} +/- {:2.4e} J".format(q,d_q))
print("time: 2 min = 120 s")
rad_flux= q/120
d_rad_flux = (1/120) * (d_q)
print("Radiant flux = {:2.4e} +/- {:2.4e} W".format(rad_flux,d_rad_flux))
#Lets assume theta aprox equal to 0
area=np.pi*(diameter/2)**2
irradiance= rad_flux/area
d_irradiance= (1/area) * d_rad_flux
print("Irradiance = {:2.4e} +/- {:2.4e} W/m**2".format(irradiance,d_irradiance))
print("Relative Error = {:2.4e} %".format(d_irradiance/irradiance*100))
print("--------------------------------------")
print("Numerical Propagation:")
num_samples=1000
energy_simul= q + d_q * np.random.randn(num_samples)
print("Radiant energy = {:2.4e} +/- {:2.4e} J".format(np.mean(energy_simul), np.std(energy_simul)))
flux_simul= energy_simul/120
mean_flux=np.mean(flux_simul) 
std_flux = np.std(flux_simul-mean_flux)
print("Irradiance = {:2.4e} +/- {:2.4e} W/m**2".format(irradiance,d_irradiance))
print("Radiant flux = {:2.4e} +/- {:2.4e} W".format(mean_flux, std_flux))
irradiance_simul=flux_simul/area
mean_irradiance=np.mean(irradiance_simul)
std_irradiance=np.std(irradiance_simul-mean_irradiance)
print("Irradiance = {:2.4e} +/- {:2.4e} W/m**2".format(mean_irradiance, std_irradiance))
print("Relative Error = {:2.4e} %".format(std_irradiance/mean_irradiance*100))
```
